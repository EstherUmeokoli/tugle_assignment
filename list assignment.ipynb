```python
t= [1,2,3]
def cumsum(t):
    d=[] #a new list is assigned to a variable d
    total= 0 #since what we re working on is an int d lowest is 0 so we re counting from 0
    
    for x in t: # for the items in the list t
        total=total+x # pick one add to the nxt one and return it like 1+2 return 3 + 3=return 6
        d.append(total) # append the total to the new list
    return d # return the cumulative total inside the new list

f= cumsum(t)
print(f)
```

```python
t= [[1,2],[3],[4,5,6]]
def nested_sum(t):
    sum=0 # we re working on integer starting from 0
    for x in t: # for the items in the list t
        if type(x)==  list:#if the data type of the item in the list is a list
            for a in x:# for the item in a in x
                sum+=a # sum it
        else:
            sum+=x # else just sum the item x in the list like dat if its not a list
    return sum# return the sum

nested_sum(t)# call the function
```
