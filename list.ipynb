```python
t= [1,2,3,4,5]
def add_up(t):
    total= 0
    for x in t:
        total+=x
    return total

f= add_up(t)
print(f)
        
```

```python
t= 'esther'
def capitalize_all(t):
    res=[]
    for s in t:
        res.append(s.capitalize())
    return res

g= capitalize_all(t)
print(g)
```

```python
t= 'EsTheR'
def only_upper(t):
    res=[]
    for s in t:
        if s.isupper():
            res.append(s)
    return res

g= only_upper(t)
print(g)
```
