```python
time_left = (6*3600)+(52*60)
easy_pace = 1
time_easypace = ((8*60)+15)*easy_pace
tempo = 3
time_tempo = tempo*((7*60)+12)
final_pace = ((8*60)+15)*easy_pace
total_time = time_left + time_easypace + time_tempo + final_pace
hours = (total_time//3600)
minutes = (total_time%3600)//60
print(f"{hours} hours {minutes} minutes")
```
