```python
def countdown(n):
    if n<=0:
        print('Blastoff!')
    else:
        print(n)
        countdown(n-1)

countdown(5)
```

```python
password = input('guess password :')
if len(password)>8 and any(char.isdigit() for char in password):
    print('Strong password')
else:
    print('Weak password')
```

```python
score = int(input('write a score (0-100) :'))
if score>=70 and score<=100:
    print('A')
elif score>=60 and score<=69:
    print('B')
elif score>=50 and score<=59:
    print('C')
elif score>=45 and score<=49:
    print('D')
else:
    print('F')
```

```python
def loan_qualification(age, income):
    if age >=21 and income >=50000:
        print('you are qualified for the loan')
    else:
        print('you are not qualified for the loan')
        
loan_qualification(30,52000)
```

```python
def countdown(n):
    if n>=0:
        print(n)
        countdown(n-1)

countdown(10)
```

```python
def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n -1)
factorial(11)
```

```python
def summation(s):
    if s==0:
        return 0
    else:
        return s+summation(s-1)

summation(5)
```

```python
def power(n,x):#the function power carries parameters of a number and power number
    if x==0: #because in normal power root anything raise to the power of 0 is 1 dats while i said it should return 1
        return 1
    else:
        return n*power(n, x-1)#dis code has a pattern how dey solve it which (x**n= x*x**n-1)normally so there is d code of it. recursion occur

power(3,9)
        
        
    
```

```python
def reverse(h):
    if h=='':
        return ''
    else:
        return reverse(h[1:]) + h[0]

reverse('happy')
```
