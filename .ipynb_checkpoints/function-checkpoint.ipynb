```python
def print_lyrics():
    print("I'm a lumberjack, and I'm okay.")
    print("I sleep all night and I work all day.")

print_lyrics()    
```

```python
def right_justify(s):
    spaces = 70 - len(s)
    print(" "*spaces + s)

right_justify("monty")
```

```python
def do_twice(d,val):
    d(val)
    d(val)
def print_twice(n):
    print(n)
    print(n)

do_twice(print_twice,'wisdom')
```
