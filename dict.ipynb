```python
s= 'square'
def histogram(s):
    d = dict()# d is a new string
    for c in s:# for thge character in string s
        if c not in d:#if the character is not inthe new dict
            d[c] = 1# map it to 1
        else:
            d[c] +=1# else just add 1
    return d

histogram(s)
```

```python
v= histogram(s)#instead of creating a new dict i will use the one i called earlier which is(histogram(s))
def old_dict(s):# i defined a new function
    for c in s:
        print(c)#print c displays only keys

old_dict(v)
```

```python
v= histogram(s)
def old_dict(s):
    for c in s:
        print(s[c])#print variable key=value its also called lookup

old_dict(v)
```

```python
def invert_dict(d):
    inverse = dict()
    for key in d:
        val = d[key]
        if val not in inverse:
            inverse[val] = [key]
        else:
            inverse[val].append(key)
    return inverse

invert_dict(v)
```

```python
#Question two
def inventory():
    bread_price= int(input('enter b price :'))
    bread_qty= int(input('enter b qty :'))
    milk_price= int(input('enter m price :'))
    milk_qty= int(input('enter m qty :'))
    egg_price= int(input('enter e price :'))
    egg_qty= int(input('enter e qty :'))
    products= {
        'Bread':{'price':bread_price, 'quantity':bread_qty}, 'Milk':{'price':milk_price, 'quantity':milk_qty}, 'Egg':{'price':egg_price, 'quantity'
        :egg_qty}}
    return products

y= inventory()
print(y)
        
    
```

```python
def choice_details():
    y= inventory()
    product= input('enter the product :')
    if product in y:
        print('Product:', product)
        print('Price:', y[product]['price'])
        print('Quantity:', y[product]['quantity'])
    else:
        print('product not found')

choice_details()
        
```

```python
def added_product():
    y= inventory()
    new= input('enter the product :')
    n_price= int(input('enter the price :'))
    n_qty= int(input('enter the qty :'))
    w= {new:{'price':n_price, 'quantity':n_qty}}
    y.update(w)
    return y           
added_product()
        
```

```python
def update_stock():
    y= inventory()
    product= input('enter a product :')
    for product in y:# tends to apply through the dict
        quantity= int(input('enter the qty :'))
        y[product]['quantity']= quantity
    print('all the stock is updated')
    return y

update_stock()


                      
        
    
```

```python
def sell_product():
    y= inventory()
    product= input('enter the product :')
    qty= int(input('enter the amount :'))
    if product in y:
        if qty <= y[product]['quantity'] and qty > 0:
            Price= y[product]['price'] * qty
            y[product]['quantity'] -= qty        
            print('Product:',product)
            print('Price:',Price)
            print('left stock:',y[product]['quantity'])
        else:
            print('insufficient stock')
    else:
        print('product not found')
        
sell_product()
```
